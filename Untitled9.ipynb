```python
#Q1
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Original paragraph
text = """
Technology makes our daily lives easier and more connected.
Smartphones and computers help people communicate, learn and work efficiently.
Students can access educational resources online from anywhere.
Businesses also use technology to improve their services.
As technology continues to grow, it creates new opportunities for everyone.
"""

# 1
text_lower = text.lower()
clean_text = re.sub(r'[^\w\s]', '', text_lower)

print("------ Cleaned Text ------")
print(clean_text)

# 2
sentences = sent_tokenize(text)
words = word_tokenize(text)

print("\n------ Sentence Tokens ------")
print(sentences)

print("\n------ Word Tokens ------")
print(words)

# 3
split_words = clean_text.split()
tokenized_words = word_tokenize(text)

print("\n------ Using split() ------")
print(split_words)

print("\n------ Using word_tokenize() ------")
print(tokenized_words)

print("\n------ Comparison ------")
print("split() separates text based on spaces only.")
print("word_tokenize() separates words and punctuation using NLP rules.")

# 4
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in word_tokenize(clean_text)
    if word not in stop_words
]

print("\n------ Words After Removing Stopwords ------")
print(filtered_words)

# 5
freq_dist = FreqDist(filtered_words)

print("\n------ Word Frequency Distribution ------")
for word, freq in freq_dist.items():
    print(f"{word}: {freq}")

print("\n------ Most Common Words ------")
print(freq_dist.most_common())
```

```python
#Q2
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')


text = """
Technology makes our daily lives easier and more connected.
Smartphones and computers help people communicate, learn and work efficiently.
Students can access educational resources online from anywhere.
Businesses also use technology to improve their services.
As technology continues to grow, it creates new opportunities for everyone.
"""

# 1
words = re.findall(r'\b[a-zA-Z]+\b', text)

print("----- Words with Only Alphabets -----")
print(words)

# 2
stop_words = set(stopwords.words('english'))

filtered_words = [
    word.lower() for word in words
    if word.lower() not in stop_words
]

print("\n----- After Removing Stopwords -----")
print(filtered_words)

# 3
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\n----- Stemmed Words -----")
print(stemmed_words)

# 4
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\n----- Lemmatized Words -----")
print(lemmatized_words)

# 5
print("\n----- Comparison -----")
for original, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(f"{original:15} Stem: {stem:12} Lemma: {lemma}")
```

```python
#Q3

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

texts = [
    "Artificial intelligence is transforming healthcare and education.",
    "This smartphone has excellent battery life and camera quality.",
    "The football team won the championship after a thrilling match."
]

# 1
print("----- BAG OF WORDS -----")

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

print(bow_df)

# 2
print("\n----- TF-IDF SCORES -----")

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print(tfidf_df)

# 3
print("\n----- TOP 3 KEYWORDS FROM EACH TEXT -----")

feature_names = tfidf_vectorizer.get_feature_names_out()

for i, row in enumerate(tfidf_matrix.toarray()):
    word_scores = list(zip(feature_names, row))

    top_keywords = sorted(
        word_scores,
        key=lambda x: x[1],
        reverse=True
    )[:3]

    print(f"\nText {i+1}:")
    print(texts[i])

    print("Top 3 Keywords:")
    for word, score in top_keywords:
        print(f"{word}: {score:.4f}")
```

```python
#Q4
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Texts
text1 = """
Artificial Intelligence enables machines to learn from data and make decisions.
AI is used in healthcare, education, and business applications.
It helps automate repetitive tasks and improve efficiency.
Machine learning is one of the most popular branches of AI.
"""

text2 = """
Blockchain is a distributed digital ledger technology.
It stores transactions securely across multiple computers.
Blockchain is widely used in cryptocurrencies and supply chain management.
It improves transparency, security, and data integrity.
"""

# 1
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = text.split()
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens of Text 1:")
print(tokens1)

print("\nTokens of Text 2:")
print(tokens2)

# 2 Jaccard Similarity
set1 = set(tokens1)
set2 = set(tokens2)

intersection = len(set1.intersection(set2))
union = len(set1.union(set2))

jaccard_similarity = intersection / union

print("\nJaccard Similarity:")
print(round(jaccard_similarity, 4))

# 2 Cosine Similarity using TF-IDF
documents = [text1, text2]

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(documents)

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\nCosine Similarity:")
print(round(cosine_sim[0][0], 4))
```

```python
#Q5
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "This smartphone is excellent. The battery life is amazing and the camera quality is outstanding.",
    "The laptop performs well and is very fast. I am happy with the purchase.",
    "The headphones are average. Sound quality is okay but the battery drains quickly.",
    "This product is terrible. It stopped working after one week and customer support was unhelpful."
]

positive_reviews = []

print("------ SENTIMENT ANALYSIS RESULTS ------")

for i, review in enumerate(reviews, start=1):

    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    # Classification
    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print(f"\nReview {i}:")
    print(review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

```python
#Q6
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Training text
text = """
Artificial intelligence helps people solve problems.
AI is used in healthcare and education.
Technology improves productivity and efficiency.
"""

# 1
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

# Create sequences
input_sequences = []
tokens = tokenizer.texts_to_sequences([text])[0]

for i in range(1, len(tokens)):
    input_sequences.append(tokens[:i+1])

# Padding
max_len = max(len(x) for x in input_sequences)

input_sequences = np.array(
    pad_sequences(input_sequences, maxlen=max_len, padding='pre')
)

# Split into X and y
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# 2
model = Sequential()
model.add(Embedding(total_words, 10, input_length=max_len-1))
model.add(LSTM(50))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam')

# 3
model.fit(X, y, epochs=100, verbose=0)

# Generate text
seed_text = "artificial"

for _ in range(5):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences(
        [token_list],
        maxlen=max_len-1,
        padding='pre'
    )

    predicted = np.argmax(model.predict(token_list, verbose=0))

    for word, index in tokenizer.word_index.items():
        if index == predicted:
            seed_text += " " + word
            break

print("Generated Text:")
print(seed_text)
```
