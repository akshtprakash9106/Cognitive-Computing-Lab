```python
# Q1
import pandas as pd

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"},
]

personalized_entries = [
    {"question": "how do i update my registered mobile number",
     "answer": "You can update your mobile number in the account settings under profile.",
     "keywords": "update mobile phone account", "category": "account"},
    {"question": "where can i find exam guidelines",
     "answer": "Exam guidelines are available on the student portal under notices.",
     "keywords": "exam rules guidelines portal", "category": "general"},
]

faq_df = pd.DataFrame(fixed_entries + personalized_entries)
print(faq_df)
```

```python
#Q2
def score_query(query, df):
    results = []
    for i, row in df.iterrows():
        score = sum(kw in query.lower() for kw in row["keywords"].split())
        if score > 0:
            results.append((row["question"], row["answer"], score))
    results.sort(key=lambda x: x[2], reverse=True)
    return results

print(score_query("how to pay fee", faq_df))
```

```python
#Q3
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

print(same_category("account", faq_df))
```

```python
#Q6
def score_query_ties(query, df):
    results = []
    for i, row in df.iterrows():
        score = sum(kw in query.lower() for kw in row["keywords"].split())
        if score > 0:
            results.append((row["question"], row["answer"], score))
    if not results:
        return []
    max_score = max(r[2] for r in results)
    tied = [r for r in results if r[2] == max_score]
    return tied

# Demonstration
print("Tie case:", score_query_ties("fee", faq_df))  # matches both fee entries
print("Non-tie case:", score_query_ties("password reset", faq_df))
```
