```python
#Q2b
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
a,b = np.unique(x, return_counts=True)
elementx = (a[np.argmax(b)])
indexx = np.where(x == elementx)
print("Max occurrence is of : ",elementx)
print("Indices : ",indexx[0])
```

```python
c,d = np.unique(y, return_counts=True)
elementy = (c[np.argmax(d)])
indexy = np.where(y == elementy)
print("Max occurrence is of : ",elementy)
print("Indices : ",indexy[0])
```

```python
#Q4
akshtprakash = np.linspace(10,100,25)
print("Dimension : ",akshtprakash.ndim)
print("Shape : ",akshtprakash.shape)
print("Total Elements : ",akshtprakash.size)
print("Data Type : ",akshtprakash.dtype)
print("Total Bytes : ",akshtprakash.nbytes)
print("Transpose using reshape :",akshtprakash.reshape(25,1))
print("Transpose using T : ",akshtprakash.T)
```
